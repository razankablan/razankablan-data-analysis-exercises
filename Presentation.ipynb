```python
# Check for missing values
bitcoin_df.isnull().sum()

# View rows that contain any NaN values
bitcoin_df[bitcoin_df.isnull().any(axis=1)]
```

```python
bitcoin_df['close'].fillna(bitcoin_df['close'].mean(), inplace=True)

bitcoin_df.fillna(method='ffill', inplace=True)
```

```python
bitcoin_df.dropna(inplace=True)
```

```python
# Convert 'date' column to datetime
bitcoin_df['date'] = pd.to_datetime(bitcoin_df['date'])

# Convert numeric columns that might be strings
bitcoin_df['volume'] = pd.to_numeric(bitcoin_df['volume'], errors='coerce')
```

```python
# Check and remove duplicates
bitcoin_df.duplicated().sum()
bitcoin_df.drop_duplicates(inplace=True)
```

```python
import numpy as np

# Identify outliers using the Interquartile Range (IQR)
Q1 = bitcoin_df['close'].quantile(0.25)
Q3 = bitcoin_df['close'].quantile(0.75)
IQR = Q3 - Q1

# Filter out outliers
filtered_df = bitcoin_df[~((bitcoin_df['close'] < (Q1 - 1.5 * IQR)) | (bitcoin_df['close'] > (Q3 + 1.5 * IQR)))]
```

```python
# Select specific columns
price_data = bitcoin_df[['date', 'close']]

# Filter rows – for example, Bitcoin prices above $10,000
high_price_df = bitcoin_df[bitcoin_df['close'] > 10000]

# Apply transformations – calculate daily return
bitcoin_df['daily_return'] = bitcoin_df['close'].pct_change() * 100
```

```python
import numpy as np

# Calculate log returns using NumPy
bitcoin_df['log_return'] = np.log(bitcoin_df['close'] / bitcoin_df['close'].shift(1))

# Create a new column classifying daily movement
bitcoin_df['trend'] = np.where(bitcoin_df['daily_return'] > 0, 'Gain', 'Loss')
```

```python
# Example 1: Extract month and year for trend analysis
bitcoin_df['month'] = bitcoin_df['date'].dt.month
bitcoin_df['year'] = bitcoin_df['date'].dt.year

# Example 2: Rolling averages (e.g., 7-day moving average)
bitcoin_df['7d_avg'] = bitcoin_df['close'].rolling(window=7).mean()
```

```python
#the 'trend' column (Gain/Loss)
trend_encoded = pd.get_dummies(bitcoin_df['trend'], prefix='trend')

# Merge with original DataFrame
bitcoin_df = pd.concat([bitcoin_df, trend_encoded], axis=1)
```
